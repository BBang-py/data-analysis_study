# 조건 필터링

```python
#수면 시간이 6시간 이상인 행 데이터 뽑기

df[df["Quality of Sleep"] >= 6]
```

```python
#직업이 의사인 행 데이터 뽑기

df[df["Occupation"] == "Doctor"]
```

```python
# 성별이 남자이면서 스트레스 레벨이 5 이상인 사람

df[(df["Gender"] == "Male") & (df["Stress Level"] >= 5)]
```

```python
#직업이 의사거나 간호사인 사람중에서 수면시간이 7시간 미만인 사람
#isin(): 컬럼안에 ()안에 있는 값이 있으면 True

df[(df["Occupation"].isin(["Doctor", "Nurse"])) & (df["Sleep Duration"] < 7)]
```

```python
#나이가 40이상 50이하인 사람
#between(a,b) : a이상 b이하

df[df["Age"].between(40,50)]
```

# 그룹별 집계

## 집계함수

| 함수               | 의미                        |
| ---------------- | ------------------------- |
| `mean()`         | 평균                        |
| `median()`       | 중앙값                       |
| `sum()`          | 합계                        |
| `min()`          | 최솟값                       |
| `max()`          | 최댓값                       |
| `count()`        | **결측값을 제외한** 값의 개수        |
| `size()`         | **결측값 포함** 그룹의 행 개수       |
| `std()`          | 표준편차 — 값이 얼마나 퍼져 있는지      |
| `var()`          | 분산 — 표준편차의 제곱             |
| `nunique()`      | 서로 다른 값의 개수, 기본적으로 결측값 제외 |
| `quantile(0.25)` | 25% 분위수                   |


```python
#직업별 사람 수/* 추가로 오름차순으로 정렬

df.groupby("Occupation").size().sort_values(ascending=True)
```

```python
#직업별 나이 평균 /* 추가로 round()로 반올림하고 오름차순으로 정렬

df.groupby("Occupation")["Age"].mean().round(1).sort_values(ascending=True)
```

```python
#BMI 범주별 스트레스 레벨과 수면 시간의 평균

df.groupby("BMI Category")[["Stress Level","Sleep Duration"]].mean().round(2)
```

```python
#집계 함수를 한번에 쓰려면 agg사용
#직업별 하루 걸음수의 평균과 최댓값 

df.groupby("Occupation")["Daily Steps"].agg(["mean" , "max"]).round(1).sort_values("mean" , ascending=False)
```

# 데이터 시각화

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

## 기초 그래프 시각화
fig: 그래프를 포함한 그림 전체<br>
ax: fig안에서 실제 그래프를 그리는 영역

```python
fig , ax = plt.subplots()
x = [1,3,5,7]
y = [1,9,26,49]
ax.scatter(x,y)
plt.show()
```

```python
#fig 하나에 ax 두개 만들기(1행2열)

fig , (ax1,ax2) = plt.subplots(1,2)
x = np.array([1,3,5,7])
y = np.array([1,8,25,49])
ax1.scatter(x,y**2)
ax2.plot(x,y**2)
```

```python
#(2,2)그래프 그리기

fig , ([ax1,ax2],[ax3,ax4]) = plt.subplots(2,2)
x = np.linspace(0,np.pi,50)
y = x * np.arange(1,51)
fig.suptitle('X,Y Graphy')
ax1.plot(x,y)
ax2.plot(x,np.sin(x ** 2),'tab:orange')
ax3.plot(y,y,'tab:red')
ax4.plot(x,-y,'tab:green')
plt.show()
```

## Bar chart

```python
#우선 직업별 걸음수의 평균을 가져옴 
mean_steps = df.groupby("Occupation")["Daily Steps"].mean().astype("int64").sort_values()
mean_steps
```

```python
#가로 10인치, 세로 6인치 크기의 그림 생성
plt.Figure(figsize=(10,6))
```

```python
#막대 그래프 생성
plt.bar(mean_steps.index,mean_steps.values)
```

```python
#위의 그래프의 직업 글자가 겹쳐 세로 막대그래프로 변경
plt.barh(mean_steps.index,mean_steps.values)
```

```python
#축 이름 설점 /* rotation= :글자 회전
#grid(): 점선 격자 추가 /* axis =: 축방향 , linestyle =: 선모양 , alpha=: 투명도

plt.xlabel("Average Daily Steps",)
plt.ylabel("Occupation",rotation=0)
plt.grid(axis="x", linestyle="-", alpha=0.3)
plt.barh(mean_steps.index,mean_steps.values)
```

## Scatter plot

```python
plt.scatter(df["Sleep Duration"],df["Stress Level"],alpha=0.3)
plt.xlabel("Sleep Duration")
plt.ylabel("Stress Level",rotation=0)
plt.show()
```

```python
#ylabel이 그래프를 넘어와서 조정해야함
#labelpad= : 축과 제목 사이 간격 늘리기
#plt.tight_layout(): 여백 자동 조절

plt.scatter(df["Sleep Duration"], df["Stress Level"], alpha=0.3)

plt.xlabel("Sleep Duration")
plt.ylabel("Stress Level", rotation=0,labelpad=40)
plt.tight_layout()
plt.show()
```
