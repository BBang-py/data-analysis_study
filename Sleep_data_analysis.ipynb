# 그룹별 집계

## 집계함수

| 함수               | 의미                        |
| ---------------- | ------------------------- |
| `mean()`         | 평균                        |
| `median()`       | 중앙값                       |
| `sum()`          | 합계                        |
| `min()`          | 최솟값                       |
| `max()`          | 최댓값                       |
| `count()`        | **결측값을 제외한** 값의 개수        |
| `size()`         | **결측값 포함** 그룹의 행 개수       |
| `std()`          | 표준편차 — 값이 얼마나 퍼져 있는지      |
| `var()`          | 분산 — 표준편차의 제곱             |
| `nunique()`      | 서로 다른 값의 개수, 기본적으로 결측값 제외 |
| `quantile(0.25)` | 25% 분위수                   |


```python
#직업별 사람 수/* 추가로 오름차순으로 정렬

df.groupby("Occupation").size().sort_values(ascending=True)
```

```python
#직업별 나이 평균 /* 추가로 round()로 반올림하고 오름차순으로 정렬

df.groupby("Occupation")["Age"].mean().round(1).sort_values(ascending=True)
```

```python
#BMI 범주별 스트레스 레벨과 수면 시간의 평균

df.groupby("BMI Category")[["Stress Level","Sleep Duration"]].mean().round(2)
```

```python
#집계 함수를 한번에 쓰려면 agg사용
#직업별 하루 걸음수의 평균과 최댓값 

df.groupby("Occupation")["Daily Steps"].agg(["mean" , "max"]).round(1).sort_values("mean" , ascending=False)
```

## shapiro

```python
from scipy.stats import shapiro
```

* 영가설(Null hypothesis): 주어진 데이터는 정규분포를 따른다
* 제1종 오류확률(p-value)이 5% 이하일 때, 즉 영가설이 기각되면 정규분포를 따르지 않습니다.

```python
a , b = shapiro(df["Age"].dropna())
print(f"Shapiro W-statistic (Normality):{a.round(5)}")
print(f"p-value: {b.round(5)}")
```

```python
a , b = shapiro(df["Sleep Duration"].dropna())
print(f"Shapiro W-statistic (Normality):{a.round(5)}")
print(f"p-value: {b.round(5)}")
```

```python
a , b = shapiro(df["Stress Level"].dropna())
print(f"Shapiro W-statistic (Normality):{a.round(5)}")
print(f"p-value: {b.round(5)}")
```

## Correlation

```python
# 귀무가설: 두 변수 사이에 선형 상관관계가 없다
# 대립가설: 두 변수 사이에 선형 상관관계가 있다
from scipy.stats import pearsonr
```

```python
a , b = pearsonr(df["Sleep Duration"], df["Stress Level"])
print(f"Pearson corr: {a.round(5)}")
print(f"p-value: {b.round(5)}\n")
if a < 0:
    print("두 데이터는 음의 방향으로 상관관계")
else:
    print("두 데이터는 양의 방향으로 상관관계")
```

## One-sample T-test

```python
# 귀무가설: 표본의 기대값(평균)은 주어진 모집단 평균과 같다
# 대립가설: 표본의 평균은 모집단 평균과 다르다
from scipy.stats import ttest_1samp
```

```python
mean_Stress_Level = df["Stress Level"].mean()
a , b = ttest_1samp(df["Stress Level"] , 5)
print(f"Stress Level의 평균:{mean_Stress_Level.round(5)}")
print(a,b)
```

## Independent Two-sample T-test

```python
from scipy.stats import ttest_ind
from scipy.stats import levene
```

```python
Sleep_Apena = df[df["Sleep Disorder"] == "Sleep Apnea"]
Insomnia = df[df["Sleep Disorder"] == "Insomnia"]
```

```python
print(f"Sleep Apena인 사람의 수면 질 점수 평균: {Sleep_Apena["Quality of Sleep"].mean()}")
print(f"Insomnia인 사람의 수면 질 점수 평균: {Insomnia["Quality of Sleep"].mean()}")
```

```python
# 두 집단의 평균을 비교하기 전에는 등분산검정을 하는것이 바람직
a , b = levene(Sleep_Apena["Quality of Sleep"],Insomnia["Quality of Sleep"])
print(a,b)

if b < 0.05:
    print("\n두 집단의 등분산성 불만족")
else:
    print("\n두 집단의 등분산성 만족")
```

```python
# equal_var = False: 두 집단의 등분산성이 불만족했을때 만족한다로 가정하는 함수

a , b = ttest_ind(Sleep_Apena["Quality of Sleep"],Insomnia["Quality of Sleep"],equal_var=False)
print(a,b)
```

# 비모수검정
## Chi-square test

```python
from scipy.stats import chi2_contingency
# 한 집단의 분포가 다른 집단의 분포와 독립적인지(동일한지) 확인합니다.
```

```python
hh = len(df[(df["Physical Activity Level"] >= df["Physical Activity Level"].mean()) & \
        (df["Stress Level"] >= df["Stress Level"].mean())])
hl = len(df[(df["Physical Activity Level"] >= df["Physical Activity Level"].mean()) & \
        (df["Stress Level"] < df["Stress Level"].mean())])
lh = len(df[(df["Physical Activity Level"] < df["Physical Activity Level"].mean()) & \
        (df["Stress Level"] >= df["Stress Level"].mean())])
ll = len(df[(df["Physical Activity Level"] < df["Physical Activity Level"].mean()) & \
        (df["Stress Level"] < df["Stress Level"].mean())])

chisq, p, dof, expected = chi2_contingency([[hh, hl],
                                        [lh, ll]])
print(f"hh인원 수 : {hh} , hl인원 수 : {hl},\nlh인원 수 : {lh} , ll인원 수: {ll}")
print(f"기대빈도 표:\n{expected.round(2)}")
#chisq : 카이제곱 통계량
#p : p-value
#dof : 자유도 ((2-1) * (2-1) = 1)
# expected = 기대빈도 표

print(chisq , p , dof,sep="\n")
```
