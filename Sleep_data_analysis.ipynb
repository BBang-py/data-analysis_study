# 조건 필터링

```python
#수면 시간이 6시간 이상인 행 데이터 뽑기

df[df["Quality of Sleep"] >= 6]
```

```python
#직업이 의사인 행 데이터 뽑기

df[df["Occupation"] == "Doctor"]
```

```python
# 성별이 남자이면서 스트레스 레벨이 5 이상인 사람

df[(df["Gender"] == "Male") & (df["Stress Level"] >= 5)]
```

```python
#직업이 의사거나 간호사인 사람중에서 수면시간이 7시간 미만인 사람
#isin(): 컬럼안에 ()안에 있는 값이 있으면 True

df[(df["Occupation"].isin(["Doctor", "Nurse"])) & (df["Sleep Duration"] < 7)]
```

```python
#나이가 40이상 50이하인 사람
#between(a,b) : a이상 b이하

df[df["Age"].between(40,50)]
```
